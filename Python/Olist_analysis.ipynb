# Data Cleaning Strategy

```python
Before Cleaning the Dataset,Each table was inspected for missing values,duplicates and data quality issues.
1.Customers  ==> No missing values or duplicates,No cleaning required.
2.orders     ==> missing values were found in date columns,these missing values are expected because some orders were cancelled or were never delivered 
                 therefore they will not be removed for delivery related analysis,only delivered orders will be considered.
3.order items==> No missing values or duplicates,No cleaning required.
4.products   ==> A total of 610 missing values belong to the same set of products so they will not be removed at this stage. The decision will be based on the
                 analysis requirements.
5.reviews    ==> Missing Values in title and message are expected because many customers provide only rating ,These columns will not be filled or removed unless
                 text analysis is performed.
6.Geolocation ==> Duplicate records are expected because the same ZIP code or location can appear multiple times deduplication will only be performed when
                  if required for geolocation analysis.
```

```python
""" Before EDA ,first convert all Date columns into Datetime format."""
```

```python
orders["order_purchase_timestamp"]=pd.to_datetime(orders["order_purchase_timestamp"])
orders["order_approved_at"]=pd.to_datetime(orders["order_approved_at"])
orders["order_delivered_carrier_date"]=pd.to_datetime(orders["order_delivered_carrier_date"])
orders["order_delivered_customer_date"]=pd.to_datetime(orders["order_delivered_customer_date"])
orders["order_estimated_delivery_date"]=pd.to_datetime(orders["order_estimated_delivery_date"])
orders.info()
```

```python
# 1.Total orders ??
total_orders=orders["order_id"].nunique()
print(total_orders)
```

```python
# 2.Order status distribution.
orders.order_status.value_counts()
```

```python
""" Business Insight """
1. Around 97% orders successfully delivered.
2. cancelled and unavailable orders are very less in numbers.
3. some orders are in processing or shipping stage.
```

```python
# 3.Monthly Order Trend
orders["Year_Month"]=orders["order_purchase_timestamp"].dt.to_period("M")
monthly_orders=orders.groupby("Year_Month")["order_id"].count()
monthly_orders
```

```python
# 4.Total Revenue
revenue_df=orders.merge(order_items,on="order_id")
Total_revenue=revenue_df["price"].sum()
print(f"Total_revenue:{Total_revenue:,.2f}")
```

```python
# 5.Monthly Revenue Trend.
Monthly_Revenue=revenue_df.groupby("Year_Month")["price"].sum()
Monthly_Revenue
```

```python
# 6. Top 10 Product Categories by revenue.
Category_revenue=order_items.merge(products[["product_id","product_category_name"]],on="product_id",how="left")
Category_revenue=Category_revenue.merge(category,on="product_category_name")
Category_revenue.groupby("product_category_name_english")["price"].sum().sort_values(ascending=False).head(10)
```

```python
# 7.Top 10 cities by Number of Orders
city_orders=orders.merge(customers[["customer_id","customer_city"]],on="customer_id",how="left")
city_orders.groupby("customer_city")["order_id"].count().sort_values(ascending=False).head(10)
```

```python
# 8.Top 10 states by Number of Orders
state_orders=orders.merge(customers[["customer_id","customer_state"]],on="customer_id",how="left")
state_orders.groupby("customer_state")["order_id"].count().sort_values(ascending=False).head(10)
```
